for 的核心是 Iterator Protocol（iter / next / StopIteration）；而 while 的核心是條件判斷與跳躍（conditional jump）。在 CPython 中，while 並不是透過 next() 實作的，而是編譯成對應的 bytecode 跳躍指令來完成迴圈。

## `for` 迴圈與 `while` 迴圈的對應

Python 的 `for` 迴圈：

```python
for variable in iterable:
    body
```

其底層邏輯可以表示為：

```python
iterator = iter(iterable)

while True:
    try:
        variable = next(iterator)
        body
    except StopIteration:
        break
```

這兩段程式碼是等價的（equivalent），只是 `for` 迴圈將許多細節隱藏起來，使程式更容易閱讀與撰寫。

## 相關名詞說明

### 1. Iterable（可迭代物件）

**Iterable** 是指可以逐一產生元素的物件；就是可以一個一個拿出資料來看的物件。

常見的 iterable：range(1, 6)、[10, 20, 30]、"Python"、{"Tom", "Mary", "John"}、{"name": "Tom", "age": 18}。
例如：numbers = [10, 20, 30]，numbers就是一個 iterable。

### 2. `iter()`

`iter()` 函式會根據 iterable 建立一個 iterator。

例如：

```python
numbers = [10, 20, 30]
iterator = iter(numbers)
```
可以理解為：Iterable  --iter()-->  Iterator

### 3. Iterator（迭代器）

Iterator 會記錄目前走訪到哪一個元素。

例如：
```python
numbers = [10, 20, 30]
iterator = iter(numbers)
```
此時 iterator 會依序提供：10 → 20 → 30。

### 4. `next()`

`next()` 用來取得 iterator 的下一個元素。

例如：

```python
numbers = [10, 20, 30]
iterator = iter(numbers)
print(next(iterator))
```
輸出10，再次呼叫：print(next(iterator))，輸出20，再次呼叫：print(next(iterator))，輸出：30。

### 5. `StopIteration`

當 iterator 沒有元素可以提供時，`next()` 會拋出（raise）一個例外：StopIteration。例如：

```python
numbers = [10, 20]
iterator = iter(numbers)
print(next(iterator))
print(next(iterator))
print(next(iterator))
```
結果：
```text
10
20
StopIteration
```

因此在底層實作中：

```python
while True:
    try:
        variable = next(iterator)
        body
    except StopIteration:
        break
```
是利用 `StopIteration` 來判斷何時結束迴圈。

# 蒙特卡洛模擬（Monte Carlo Simulation）

蒙特卡洛模擬（Monte Carlo simulation）使用亂數及機率來解決問題。  
這個方法廣泛使用於計算數學、物理、化學及財經上。介紹一個使用蒙特卡洛模擬方法來預估 $\pi$ 的例子。

要使用蒙特卡洛方法預估$\pi$，請繪製一個帶有邊界方框的圓形，如下所示：

<p align="center">
  <img src="./images/fig5.6.3.png" width="300"><br>
  <em></em>
</p>


假設圓半徑為 1，那麼圓面積就是$\pi$，而邊界方框的面積為 4。隨機在邊界方框內產生一個點，該點落於圓內的機率是circleArea / squareArea = $\pi / 4$。

請撰寫一程式，隨機產生邊界方框內的 1,000,000 個點，並以 `numberOfHits` 表示落於圓內的點個數。因此，numberOfHits 約為 $1000000 \times (\pi / 4)$。$\pi$ 約為 $4 \times$ numberOfHits / 1000000。完整程式如範例程式 5.10 所示。





````python
# Introduction to Programming Using Python Example 5.10 MonteCarloSimulation.py
import random


NUMBER_OF_TRIALS = 100000 # Constant
numberOfHits = 0

for i in range(NUMBER_OF_TRIALS):
    x = random.random() * 2 - 1
    y = random.random() * 2 - 1


    if x * x + y * y <= 1:
        numberOfHits += 1


pi = 4 * numberOfHits / NUMBER_OF_TRIALS


print("PI is", pi)
````
