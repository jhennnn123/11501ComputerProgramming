1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
將十進位小數轉成二進位的方法是：

小數部分持續乘以 2，每次取整數部分作為下一個二進位數字。

計算過程
步驟	計算結果	取整數部分
1	0.1 × 2 = 0.2	0
2	0.2 × 2 = 0.4	0
3	0.4 × 2 = 0.8	0
4	0.8 × 2 = 1.6	1
5	0.6 × 2 = 1.2	1
6	0.2 × 2 = 0.4	0
7	0.4 × 2 = 0.8	0
8	0.8 × 2 = 1.6	1
...	...	...
因此得到：

0.1(10)
=
0.0001100110011001100110011...(2)

其中:0011會一直重複出現。

理由:
因為 0.1 在十進位中可以有限表示，但轉換成二進位後會形成循環小數，所以無法用有限個二進位位元精確表示。

現象:
可以發現 0.1 的二進位表示法是一個無限循環小數，因此電腦只能儲存最接近的近似值，而無法精確儲存真正的 0.1。


2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
Python 的浮點數使用 IEEE 754 雙精度浮點數（Double Precision）。

由於電腦記憶體有限，無法儲存無限多位數，因此會：

只保留有限的二進位位數。
超出的部分進行四捨五入。
採用 IEEE 754 的捨入規則，也就是「四捨六入五成雙」。

因此 Python 儲存的不是精確的 0.1，而是一個非常接近 0.1 的近似值。

若只是讓結果顯示漂亮，使用 round() 可以接受；但它只是把誤差隱藏起來，並沒有真正消除誤差。

在科學計算中，通常使用 IEEE 754 浮點數，因為運算速度快，而且微小誤差通常可以接受。

在金融計算中，金額必須精確，因此較適合使用 Decimal，避免長期累積誤差。
