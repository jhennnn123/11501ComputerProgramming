### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python int：追求數值完全正確
FP16/BF16：追求訓練與推論效率
2. 各自的優點與缺點。
Python 的 int

優點：

整數運算結果精確
幾乎不會發生整數溢位
適合金融、密碼學、科學計算等需要高正確性的領域

缺點：

數值越大，占用越多記憶體
運算速度較慢
不利於大量平行運算

FP16、BF16

優點：

記憶體需求小
GPU 運算速度快
可儲存更多模型參數
降低訓練成本

缺點：

存在捨入誤差
精度較低
極小或極大的數值可能造成數值不穩定
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
大型語言模型通常選擇較低精度（如 FP16、BF16）的表示方式，因為可以大幅降低記憶體使用量、提升 GPU 運算速度，並在維持足夠準確度的同時提高訓練與推論效率。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
(1) 記憶體需求大幅增加
(2) 訓練速度變慢
(3) 可訓練模型規模變小
(4) 成本上升

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
Python 可以執行 10**1000，是因為 int 採用任意精度設計，數字變大時會自動使用更多記憶體來儲存；但 AI 模型通常有數十億個權重，如果每個數字都使用高精度表示，會占用大量記憶體、降低運算速度，且 GPU 對 FP16、BF16、FP32 有專門加速，因此 AI 會選擇較低精度的格式，在維持足夠準確度的同時獲得更快的訓練與推論效率。
